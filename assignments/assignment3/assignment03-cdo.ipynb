#### Goal: 
- This program retrieves the dataset for the **"exchequer account (historical series)"** from the CSO, and stores it into a file called "cso.json".

Note, the direct link to the source of the data on the CSO website can be found [here](https://data.cso.ie/table/FIQ02).


```python
# import requests library to interact with API
import requests

# import json library to work with JSON data
import json
```

```python
# define the API endpoint URL and make a GET request to the API for the status code
url = "https://ws.cso.ie/public/api.jsonrpc?data=%7B%22jsonrpc%22:%222.0%22,%22method%22:%22PxStat.Data.Cube_API.ReadDataset%22,%22params%22:%7B%22class%22:%22query%22,%22id%22:%5B%5D,%22dimension%22:%7B%7D,%22extension%22:%7B%22pivot%22:null,%22codes%22:false,%22language%22:%7B%22code%22:%22en%22%7D,%22format%22:%7B%22type%22:%22JSON-stat%22,%22version%22:%222.0%22%7D,%22matrix%22:%22FIQ02%22%7D,%22version%22:%222.0%22%7D%7D"
response = requests.get(url)

print (response.status_code)
```

```python
# save the response to a variable and print it out
exchange_rate_data = response.json()
# print(exchange_rate_data)
```

```python
# save the response to a json file
with open('cso.json', 'w') as json_file:
    json.dump(exchange_rate_data, json_file, indent=2)   
```
